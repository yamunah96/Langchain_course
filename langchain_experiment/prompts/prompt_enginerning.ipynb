```python
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
llm= ChatOpenAI(
    model="openai/gpt-oss-120b",
    api_key=os.getenv('GROQ_API_KEY'),
    base_url= "https://api.groq.com/openai/v1",
    max_tokens= 1000
)
    
```

```python
llm
```

```python
prompts=[
    ("system","You are a python developer"),
    ("user","How to sort the array")
]
response= llm.invoke(prompts)
print(response.content)
```

Prompt Templates-Dynamic Prompts

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
out=StrOutputParser()
```

```python
prompts=ChatPromptTemplate.from_messages([
   {"role":"system","content":"You are translator and translate the sentences to {language}"},
   {"role":"user","content":"{query}"} 
])
prompts
```

```python
final_propmts=prompts.format_messages(
    language="kannada",
    query="I Love Python"
)
res=llm.invoke(final_propmts)
res.content
```

```python
#### Chains- Runnable
# prompts.invoke({'language':"kannada","query":"i love karnataka"})
res= llm.invoke(prompts.invoke({'language':"kannada","query":"i love karnataka"}))
res.content
```

```python
#### My Self Chain Node

def transform_case(result:str):
    return result.upper()
```

```python
chains=prompts | llm | out| transform_case
res=chains.invoke({"language":"Hindi", "query":"I love LangChain ?"})
print(res)
```
